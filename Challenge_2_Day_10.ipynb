```python
from pyspark.sql.functions import col
import time

start = time.time()

purchase_df = spark.sql("""
SELECT * 
FROM ecommerce.events_delta
WHERE event_type = 'purchase'
""")

purchase_df.count()

end = time.time()

print("Execution Time (Without Optimization):", end - start, "seconds")
```

```python
purchase_df.explain(True)
```

```python
%sql
DESCRIBE DETAIL ecommerce.events_delta
```

```python
df = spark.table("ecommerce.events_delta")
```

```python
import time
from pyspark.sql.functions import col

start = time.time()

df.filter(col("event_type") == "purchase").count()

end = time.time()
print("Without Cache:", end - start, "seconds")
```

```python
%sql
OPTIMIZE ecommerce.events_delta;
```

```python
%sql
OPTIMIZE ecommerce.events_delta
ZORDER BY (event_type);
```

```python
start = time.time()

spark.sql("""
SELECT *
FROM ecommerce.events_delta
WHERE event_type = 'purchase'
""").count()

end = time.time()

print("Execution Time After OPTIMIZE:", end - start, "seconds")
```
