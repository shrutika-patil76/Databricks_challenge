```python
from pyspark.sql import functions as F

# Load tables
events = spark.read.table("default.bronze_events")
features_df = spark.read.table("default.silver_user_features")
```

```python
label_df = events.groupBy("user_id") \
    .agg(
        F.max(
            F.when(F.col("event_type") == "purchase", 1).otherwise(0)
        ).alias("purchased")
    )
```

```python
training_data = features_df.join(label_df, "user_id", "left")
training_data = training_data.fillna({"purchased": 0})
```

```python
class_counts = training_data.groupBy("purchased").count().collect()

count_dict = {row["purchased"]: row["count"] for row in class_counts}
total = sum(count_dict.values())

training_data = training_data.withColumn(
    "class_weight",
    F.when(F.col("purchased") == 1,
           total / (2 * count_dict[1]))
     .otherwise(total / (2 * count_dict[0]))
)
```

```python
train, test = training_data.randomSplit([0.8, 0.2], seed=42)
```

```python
from pyspark.ml.feature import VectorAssembler

feature_columns = [
    "total_events",
    "total_spent",
    "avg_price"
]

assembler = VectorAssembler(
    inputCols=feature_columns,
    outputCol="features"
)

train_data = assembler.transform(train)
test_data = assembler.transform(test)

train_data = train_data.select("features","purchased","class_weight")
test_data = test_data.select("features","purchased","class_weight")
```

```python
from pyspark.ml.classification import LogisticRegression

lr = LogisticRegression(
    featuresCol="features",
    labelCol="purchased",
    weightCol="class_weight",
    maxIter=20
)

lr_model = lr.fit(train_data)
lr_predictions = lr_model.transform(test_data)
```

```python
from pyspark.ml.classification import RandomForestClassifier

rf = RandomForestClassifier(
    featuresCol="features",
    labelCol="purchased",
    weightCol="class_weight",
    numTrees=50,
    maxDepth=5
)

rf_model = rf.fit(train_data)
rf_predictions = rf_model.transform(test_data)
```

```python
from pyspark.ml.evaluation import BinaryClassificationEvaluator

evaluator = BinaryClassificationEvaluator(
    labelCol="purchased",
    rawPredictionCol="rawPrediction",
    metricName="areaUnderROC"
)

lr_auc = evaluator.evaluate(lr_predictions)
rf_auc = evaluator.evaluate(rf_predictions)

print(" Logistic Regression AUC:", lr_auc)
print(" Random Forest AUC:", rf_auc)
```

```python
from pyspark.ml.classification import RandomForestClassifier
from pyspark.ml.evaluation import BinaryClassificationEvaluator

evaluator = BinaryClassificationEvaluator(
    labelCol="purchased",
    rawPredictionCol="rawPrediction",
    metricName="areaUnderROC"
)

numTrees_list = [20, 50]
maxDepth_list = [3, 5]

best_auc = 0
best_params = None
best_model = None

for n in numTrees_list:
    for d in maxDepth_list:
        
        rf = RandomForestClassifier(
            featuresCol="features",
            labelCol="purchased",
            weightCol="class_weight",
            numTrees=n,
            maxDepth=d
        )
        
        model = rf.fit(train_data)
        predictions = model.transform(test_data)
        auc = evaluator.evaluate(predictions)
        
        print(f"numTrees={n}, maxDepth={d}, AUC={auc}")
        
        if auc > best_auc:
            best_auc = auc
            best_params = (n, d)
            best_model = model

print("\nBest AUC:", best_auc)
print("Best Parameters -> numTrees:", best_params[0], 
      ", maxDepth:", best_params[1])
```

```python
import mlflow
import random
import time

# Function to simulate model training
def run_experiment(num_trees, max_depth):

    with mlflow.start_run():

        # Log parameters
        mlflow.log_param("model_type", "RandomForest")
        mlflow.log_param("numTrees", num_trees)
        mlflow.log_param("maxDepth", max_depth)

        # Simulate training delay
        time.sleep(1)

        # Simulate evaluation metric
        auc_score = round(random.uniform(0.75, 0.95), 4)

        # Log metric
        mlflow.log_metric("AUC", auc_score)

        print(f"Run Completed | Trees={num_trees}, Depth={max_depth}, AUC={auc_score}")


# Run multiple experiments (like hyperparameter tuning)
run_experiment(10, 5)
run_experiment(20, 7)
run_experiment(30, 10)

print("\nAll runs logged successfully!")
```

```python
import mlflow
import mlflow.spark
from pyspark.ml.classification import LogisticRegression
from pyspark.ml.evaluation import BinaryClassificationEvaluator

# End any active run first
mlflow.end_run()

# Evaluator
evaluator = BinaryClassificationEvaluator(
    labelCol="purchased",
    metricName="areaUnderROC"
)

def run_logistic_reg(max_iter):

    with mlflow.start_run():

        # Define model
        lr = LogisticRegression(
            featuresCol="features",
            labelCol="purchased",
            weightCol="class_weight",
            maxIter=max_iter
        )

        # Train model
        lr_model = lr.fit(train_data)

        # Predictions
        predictions = lr_model.transform(test_data)

        # Evaluate
        auc = evaluator.evaluate(predictions)

        # Log parameters
        mlflow.log_param("model_type", "LogisticRegression")
        mlflow.log_param("maxIter", max_iter)

        # Log metric
        mlflow.log_metric("AUC", auc)

        print(f"Run Complete | maxIter={max_iter} | AUC={auc}")


# Run multiple configurations
run_logistic_reg(10)
run_logistic_reg(20)
run_logistic_reg(30)
run_logistic_reg(50)
run_logistic_reg(80)

print("\nAll Logistic Regression runs logged successfully!")
```

```python
import mlflow
from pyspark.ml.classification import RandomForestClassifier
from pyspark.ml.evaluation import BinaryClassificationEvaluator

# End any active MLflow run
mlflow.end_run()

# Evaluator
evaluator = BinaryClassificationEvaluator(
    labelCol="purchased",
    metricName="areaUnderROC"
)

def run_random_forest(num_trees, max_depth):

    with mlflow.start_run():

        # Define model
        rf = RandomForestClassifier(
            featuresCol="features",
            labelCol="purchased",
            weightCol="class_weight",
            numTrees=num_trees,
            maxDepth=max_depth
        )

        # Train
        rf_model = rf.fit(train_data)

        # Predict
        predictions = rf_model.transform(test_data)

        # Evaluate
        auc = evaluator.evaluate(predictions)

        # Log parameters
        mlflow.log_param("model_type", "RandomForest")
        mlflow.log_param("numTrees", num_trees)
        mlflow.log_param("maxDepth", max_depth)

        # Log metric
        mlflow.log_metric("AUC", auc)

        print(f"Run Complete | Trees={num_trees}, Depth={max_depth}, AUC={auc}")


# Multiple hyperparameter runs
run_random_forest(20, 5)
run_random_forest(50, 5)
run_random_forest(50, 10)
run_random_forest(100, 10)
run_random_forest(150, 15)

print("\nAll Random Forest runs logged successfully!")
```
