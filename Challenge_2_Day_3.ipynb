```python
# Create widget
dbutils.widgets.text("input_month", "2019-Oct")

# Get widget value
input_month = dbutils.widgets.get("input_month")

print("Processing:", input_month)
```

```python
df = spark.read.format("csv") \
    .option("header", "true") \
    .option("inferSchema", "true") \
    .load(f"/Volumes/workspace/ecommerce/ecommerce_data/{input_month}.csv")

display(df)
```

```python
df.printSchema()
```

```python
from pyspark.sql.functions import col, year, month as spark_month, dayofmonth

def create_features(df):
    df = df.withColumn("year", year(col("event_time")))
    df = df.withColumn("month", spark_month(col("event_time")))
    df = df.withColumn("day", dayofmonth(col("event_time")))
    df = df.withColumn("price_bucket", (col("price") / 100).cast("int"))
    return df
```

```python
df = create_features(df)
display(df)
```

```python
df.write.mode("overwrite").format("delta").saveAsTable("ecommerce_processed_day3")
```
