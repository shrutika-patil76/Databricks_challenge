```python
import time

start = time.time()

df = spark.table("ecommerce.events_delta") \
          .filter("event_type = 'purchase'") \
          .groupBy("brand") \
          .count()

df.show()

end = time.time()

print("Execution Time:", end - start, "seconds")
```

```python
spark.table("ecommerce.events_delta") \
     .filter("event_type = 'purchase'") \
     .groupBy("brand") \
     .count() \
     .explain(True)
```

```python
bad_df = spark.table("ecommerce.events_delta") \
              .filter("event_type = 'purchase'")

# Avoid doing this in real projects
bad_df.count()
bad_df.show()
bad_df.collect()
```

```python
spark.table("ecommerce.events_delta") \
    .write \
    .format("delta") \
    .mode("overwrite") \
    .partitionBy("event_type") \
    .saveAsTable("ecommerce.events_partitioned")
```

```python
spark.table("ecommerce.events_partitioned") \
     .filter("event_type = 'purchase'") \
     .explain(True)
```

```python
start = time.time()

optimized_df = spark.table("ecommerce.events_partitioned") \
                    .filter("event_type = 'purchase'") \
                    .groupBy("brand") \
                    .count()

optimized_df.show()

end = time.time()

print("Execution Time (Partitioned Table):", end - start, "seconds")
```
